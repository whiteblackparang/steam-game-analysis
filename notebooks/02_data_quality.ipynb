```python
import pandas as pd
from pathlib import Path

base_dir = Path.cwd().parent
raw_dir = base_dir / "data" / "raw"

games = pd.read_csv(raw_dir / "games.csv")
users = pd.read_csv(raw_dir / "users.csv")

recommendations_sample = pd.read_csv(
    raw_dir / "recommendations.csv",
    nrows=100_000
)
```

#### 결측치 확인

```python
print("[games]")
print(games.isna().sum())

print("\n[users]")
print(users.isna().sum())

print("\n[recommendations sample]")
print(recommendations_sample.isna().sum())
```

#### 중복 확인

```python
print("games app_id duplicates:",
      games["app_id"].duplicated().sum())

print("users user_id duplicates:",
      users["user_id"].duplicated().sum())

print("recommendations review_id duplicates:",
      recommendations_sample["review_id"].duplicated().sum())
```

```python
print("games duplicate rows:", games.duplicated().sum())
print("users duplicate rows:", users.duplicated().sum())
print("recommendations sample duplicate rows:",
      recommendations_sample.duplicated().sum())
```

#### 전체 행 중복 확인

```python
print("\npositive_ratio")
print(games["positive_ratio"].describe())

print("\nprice_final")
print(games["price_final"].describe())

print("\ndiscount")
print(games["discount"].describe())
```

#### 게임 데이터 값 확인

```python
print(games["rating"].value_counts(dropna=False))
```

```python
print("[positive_ratio]")
print(games["positive_ratio"].describe())

print("\n[price_final]")
print(games["price_final"].describe())

print("\n[price_original]")
print(games["price_original"].describe())

print("\n[discount]")
print(games["discount"].describe())
```

#### 게임 플랫폼/Steam Deck 값 확인

```python
print("[win]")
print(games["win"].value_counts())

print("\n[mac]")
print(games["mac"].value_counts())

print("\n[linux]")
print(games["linux"].value_counts())

print("\n[steam_deck]")
print(games["steam_deck"].value_counts())
```

#### 추천 데이터 값 확인

#### User 데이터 값 확인

```python
print("[products]")
print(users["products"].describe())

print("\n[reviews]")
print(users["reviews"].describe())
```

#### 날짜 데이터 확인

```python
games_release_date = pd.to_datetime(
    games["date_release"],
    errors="coerce"
)

recommendation_date = pd.to_datetime(
    recommendations_sample["date"],
    errors="coerce"
)

print("games invalid dates:", games_release_date.isna().sum())
print("recommendations invalid dates:",
      recommendation_date.isna().sum())
```

```python
print("Game release date range:")
print(games_release_date.min(), "~", games_release_date.max())

print("\nRecommendation date range:")
print(recommendation_date.min(), "~", recommendation_date.max())
```

####  데이터 연결성 확인

```python
app_id_match = recommendations_sample["app_id"].isin(
    games["app_id"]
)

print("Matched app_id:", app_id_match.sum())
print("Unmatched app_id:", (~app_id_match).sum())
print("Match rate:", app_id_match.mean())
```

```python
user_id_match = recommendations_sample["user_id"].isin(
    users["user_id"]
)

print("Matched user_id:", user_id_match.sum())
print("Unmatched user_id:", (~user_id_match).sum())
print("Match rate:", user_id_match.mean())
```
