```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path.cwd().parent
raw_dir = base_dir / "data" / "raw"

games_path = raw_dir / "games.csv"
users_path = raw_dir / "users.csv"
recommendations_path = raw_dir / "recommendations.csv"
```

```python
games = pd.read_csv(games_path)

print(f"Shape: {games.shape}")
games.head()
```

```python
users = pd.read_csv(users_path)

print(f"Shape: {users.shape}")
users.head()
```

```python
recommendations_sample = pd.read_csv(
    recommendations_path,
    nrows=100_000
)

print(f"Sample shape: {recommendations_sample.shape}")
recommendations_sample.head()
```

```python
print("games")
games.info()
```

```python
print("\nusers")
users.info()
```

```python
print("\nrecommendations")
recommendations_sample.info()
```

```python
print("games:", games.columns.tolist())
print("users:", users.columns.tolist())
print("recommendations:", recommendations_sample.columns.tolist())
```

```python
print(
    f"games: "
    f"{games.memory_usage(deep=True).sum() / 1024**2:.2f} MB"
)

print(
    f"users: "
    f"{users.memory_usage(deep=True).sum() / 1024**2:.2f} MB"
)

print(
    f"recommendations sample: "
    f"{recommendations_sample.memory_usage(deep=True).sum() / 1024**2:.2f} MB"
)
```
